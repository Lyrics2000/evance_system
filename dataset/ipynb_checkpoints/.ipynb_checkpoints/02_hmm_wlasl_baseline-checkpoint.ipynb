# Hidden Markov Model Baseline for WLASL Sign Recognition

This notebook builds a complete **HMM baseline** for sign recognition using WLASL videos loaded through FiftyOne.

HMM is included because it is a traditional temporal model. Unlike Random Forest, HMM keeps the sign as a sequence of frame-level observations.

## What it does

1. Loads WLASL from FiftyOne/Hugging Face.
2. Selects a manageable subset of classes.
3. Extracts pose, hand, and face landmark sequences using MediaPipe.
4. Applies scaling and PCA to reduce feature size.
5. Trains one Gaussian HMM per sign class.
6. Predicts by selecting the class with the highest log-likelihood.
7. Evaluates using accuracy, precision, recall, F1-score, classification report, and confusion matrix.
8. Saves the HMM model package and metadata.

## 1. Optional installation

```python
import os
import sys
import json
import logging
from pathlib import Path
from datetime import datetime
from collections import Counter

import cv2
import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import fiftyone as fo
import fiftyone.utils.huggingface as fouh

import mediapipe as mp
from hmmlearn.hmm import GaussianHMM

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.decomposition import PCA
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    precision_recall_fscore_support,
)
```

## 3. Configure folders and logging

```python
PROJECT_DIR = Path("ksl_project_data").resolve()
LOG_DIR = PROJECT_DIR / "logs"
FEATURE_DIR = PROJECT_DIR / "features"
MODEL_DIR = PROJECT_DIR / "models"

for folder in [PROJECT_DIR, LOG_DIR, FEATURE_DIR, MODEL_DIR]:
    folder.mkdir(parents=True, exist_ok=True)

log_file = LOG_DIR / f"hmm_wlasl_{datetime.now().strftime('%Y%m%d_%H%M%S')}.log"

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s | %(levelname)s | %(message)s",
    handlers=[logging.FileHandler(log_file), logging.StreamHandler(sys.stdout)],
    force=True,
)

logger = logging.getLogger(__name__)
logger.info("HMM notebook started")
print("Project directory:", PROJECT_DIR)
```

## 4. Load WLASL dataset

```python
DATASET_NAME = "Voxel51/WLASL"

if DATASET_NAME in fo.list_datasets():
    logger.info("Loading existing local WLASL dataset")
    dataset = fo.load_dataset(DATASET_NAME)
else:
    logger.info("Loading WLASL from Hugging Face/FiftyOne hub")
    dataset = fouh.load_from_hub(DATASET_NAME, persistent=True)

dataset.persistent = True

logger.info(f"Dataset loaded: {dataset.name}")
logger.info(f"Total samples: {len(dataset)}")

print("Dataset:", dataset.name)
print("Total samples:", len(dataset))
print("Sample fields:", dataset.get_field_schema())
```

## 5. Helper function to extract labels

```python
def get_label_from_sample(sample):
    candidate_fields = ["ground_truth", "label", "labels", "classification", "word", "gloss"]

    for field in candidate_fields:
        if field in sample:
            value = sample[field]
            if value is None:
                continue
            if hasattr(value, "label"):
                return value.label
            if isinstance(value, str):
                return value

    for field in sample.field_names:
        value = sample[field]
        if hasattr(value, "label"):
            return value.label
        if field.lower() in ["word", "gloss", "label"] and isinstance(value, str):
            return value

    return None

labels = []
missing_label_count = 0

for sample in dataset:
    label = get_label_from_sample(sample)
    if label is None:
        missing_label_count += 1
    else:
        labels.append(label)

label_counts = Counter(labels)

print("Samples with labels:", len(labels))
print("Samples missing labels:", missing_label_count)
print("Number of unique labels:", len(label_counts))
print("Top 20 labels:")
for label, count in label_counts.most_common(20):
    print(label, count)
```

## 6. Select a manageable subset

```python
N_CLASSES = 8
MIN_SAMPLES_PER_CLASS = 5
MAX_SAMPLES_PER_CLASS = 40

eligible_labels = [label for label, count in label_counts.most_common() if count >= MIN_SAMPLES_PER_CLASS]
selected_labels = eligible_labels[:N_CLASSES]
selected_labels_set = set(selected_labels)

selected_samples = []
per_class_counter = Counter()

for sample in dataset:
    label = get_label_from_sample(sample)
    if label in selected_labels_set and per_class_counter[label] < MAX_SAMPLES_PER_CLASS:
        if sample.filepath and Path(sample.filepath).exists():
            selected_samples.append(sample)
            per_class_counter[label] += 1

print("Selected classes:", selected_labels)
print("Selected usable video samples:", len(selected_samples))
print("Per-class counts:", per_class_counter)
```

## 7. Extract sequential MediaPipe landmarks

For HMM, each video remains a sequence: frame 1, frame 2, frame 3, and so on.

```python
mp_holistic = mp.solutions.holistic

POSE_LANDMARKS = 33
HAND_LANDMARKS = 21
FACE_LANDMARKS = 468
FEATURES_PER_LANDMARK = 3


def flatten_landmarks(landmarks, expected_count):
    if landmarks is None:
        return np.zeros(expected_count * FEATURES_PER_LANDMARK, dtype=np.float32)

    values = []
    for lm in landmarks.landmark[:expected_count]:
        values.extend([lm.x, lm.y, lm.z])

    expected_length = expected_count * FEATURES_PER_LANDMARK
    if len(values) < expected_length:
        values.extend([0.0] * (expected_length - len(values)))

    return np.array(values[:expected_length], dtype=np.float32)


def extract_video_landmark_sequence(video_path, max_frames=60, frame_stride=2):
    cap = cv2.VideoCapture(str(video_path))
    if not cap.isOpened():
        raise ValueError(f"Could not open video: {video_path}")

    sequence = []
    frame_index = 0

    with mp_holistic.Holistic(
        static_image_mode=False,
        model_complexity=1,
        enable_segmentation=False,
        refine_face_landmarks=False,
        min_detection_confidence=0.5,
        min_tracking_confidence=0.5,
    ) as holistic:
        while cap.isOpened() and len(sequence) < max_frames:
            success, frame = cap.read()
            if not success:
                break

            if frame_index % frame_stride != 0:
                frame_index += 1
                continue

            rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
            results = holistic.process(rgb)

            pose = flatten_landmarks(results.pose_landmarks, POSE_LANDMARKS)
            left_hand = flatten_landmarks(results.left_hand_landmarks, HAND_LANDMARKS)
            right_hand = flatten_landmarks(results.right_hand_landmarks, HAND_LANDMARKS)
            face = flatten_landmarks(results.face_landmarks, FACE_LANDMARKS)

            sequence.append(np.concatenate([pose, left_hand, right_hand, face]))
            frame_index += 1

    cap.release()

    if len(sequence) == 0:
        return None

    return np.vstack(sequence).astype(np.float32)
```

## 8. Extract or load cached HMM sequences

```python
HMM_SEQUENCE_CACHE = FEATURE_DIR / "wlasl_hmm_sequences.joblib"

if HMM_SEQUENCE_CACHE.exists():
    logger.info(f"Loading cached HMM sequences from {HMM_SEQUENCE_CACHE}")
    cached = joblib.load(HMM_SEQUENCE_CACHE)
    sequences = cached["sequences"]
    sequence_labels = cached["labels"]
else:
    logger.info("Extracting HMM landmark sequences")
    sequences = []
    sequence_labels = []
    failed = []

    for index, sample in enumerate(selected_samples, start=1):
        label = get_label_from_sample(sample)
        try:
            sequence = extract_video_landmark_sequence(sample.filepath, max_frames=60, frame_stride=2)
            if sequence is None:
                failed.append((sample.filepath, "No landmarks"))
                continue

            sequences.append(sequence)
            sequence_labels.append(label)

            if index % 10 == 0:
                logger.info(f"Processed {index}/{len(selected_samples)} videos")
        except Exception as exc:
            failed.append((sample.filepath, str(exc)))
            logger.warning(f"Failed {sample.filepath}: {exc}")

    joblib.dump({"sequences": sequences, "labels": sequence_labels}, HMM_SEQUENCE_CACHE)
    logger.info(f"Saved HMM sequence cache: {HMM_SEQUENCE_CACHE}")
    logger.info(f"Failed videos: {len(failed)}")

print("Number of sequences:", len(sequences))
print("Number of labels:", len(sequence_labels))
print("Example sequence shape:", sequences[0].shape if sequences else None)
print("Classes:", sorted(set(sequence_labels)))
```

## 9. Train/test split by video sequence

```python
label_encoder = LabelEncoder()
y_encoded = label_encoder.fit_transform(sequence_labels)
indices = np.arange(len(sequences))

train_idx, test_idx, y_train, y_test = train_test_split(
    indices,
    y_encoded,
    test_size=0.25,
    random_state=42,
    stratify=y_encoded,
)

train_sequences = [sequences[i] for i in train_idx]
test_sequences = [sequences[i] for i in test_idx]

print("Training sequences:", len(train_sequences))
print("Testing sequences:", len(test_sequences))
for idx, label in enumerate(label_encoder.classes_):
    print(idx, "=>", label)
```

## 10. Scale and reduce feature dimension

HMM struggles with very high-dimensional inputs, so we use StandardScaler and PCA.

```python
N_PCA_COMPONENTS = 30

all_train_frames = np.vstack(train_sequences)

scaler = StandardScaler()
all_train_scaled = scaler.fit_transform(all_train_frames)

pca = PCA(n_components=N_PCA_COMPONENTS, random_state=42)
pca.fit(all_train_scaled)


def transform_sequence_for_hmm(sequence):
    scaled = scaler.transform(sequence)
    reduced = pca.transform(scaled)
    return reduced.astype(np.float64)

train_sequences_reduced = [transform_sequence_for_hmm(seq) for seq in train_sequences]
test_sequences_reduced = [transform_sequence_for_hmm(seq) for seq in test_sequences]

print("Original frame feature size:", train_sequences[0].shape[1])
print("Reduced frame feature size:", train_sequences_reduced[0].shape[1])
print("Explained variance ratio:", round(float(np.sum(pca.explained_variance_ratio_)), 4))
```

```python
## 11. Train one Gaussian HMM per class
```

```python
N_HMM_STATES = 5
N_ITER = 100

class_models = {}

for class_id, class_name in enumerate(label_encoder.classes_):
    class_train_sequences = [seq for seq, label in zip(train_sequences_reduced, y_train) if label == class_id]

    if len(class_train_sequences) < 2:
        logger.warning(f"Skipping class {class_name}: too few training sequences")
        continue

    X_class = np.vstack(class_train_sequences)
    lengths = [len(seq) for seq in class_train_sequences]

    model = GaussianHMM(
        n_components=N_HMM_STATES,
        covariance_type="diag",
        n_iter=N_ITER,
        random_state=42,
        verbose=False,
    )

    logger.info(f"Training HMM for class: {class_name}, sequences: {len(class_train_sequences)}")

    try:
        model.fit(X_class, lengths)
        class_models[class_id] = model
    except Exception as exc:
        logger.warning(f"Failed to train HMM for class {class_name}: {exc}")

print("Trained HMM class models:", len(class_models))
print("Classes with models:", [label_encoder.classes_[idx] for idx in class_models.keys()])
```

## 12. Predict using highest log-likelihood

```python
def predict_hmm(sequence):
    scores = {}
    for class_id, model in class_models.items():
        try:
            scores[class_id] = model.score(sequence)
        except Exception:
            scores[class_id] = -np.inf

    if not scores:
        return None

    return max(scores, key=scores.get)


y_pred = []
for seq in test_sequences_reduced:
    pred = predict_hmm(seq)
    y_pred.append(-1 if pred is None else pred)

y_pred = np.array(y_pred)
valid_mask = y_pred != -1

y_test_valid = y_test[valid_mask]
y_pred_valid = y_pred[valid_mask]

print("Valid predictions:", len(y_pred_valid), "out of", len(y_test))
```

## 13. Evaluate HMM model

```python
accuracy = accuracy_score(y_test_valid, y_pred_valid)
precision, recall, f1, _ = precision_recall_fscore_support(y_test_valid, y_pred_valid, average="weighted", zero_division=0)

print("Accuracy:", round(accuracy, 4))
print("Weighted Precision:", round(precision, 4))
print("Weighted Recall:", round(recall, 4))
print("Weighted F1-score:", round(f1, 4))

print("Classification Report:")
print(classification_report(
    y_test_valid,
    y_pred_valid,
    labels=list(range(len(label_encoder.classes_))),
    target_names=label_encoder.classes_,
    zero_division=0,
))
```

## 14. Confusion matrix

```python
cm = confusion_matrix(y_test_valid, y_pred_valid, labels=list(range(len(label_encoder.classes_))))
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=label_encoder.classes_)
fig, ax = plt.subplots(figsize=(10, 10))
disp.plot(ax=ax, xticks_rotation=45)
plt.title("HMM Confusion Matrix")
plt.show()
```

## 15. Save HMM package and metadata

```python
hmm_package_path = MODEL_DIR / "hmm_wlasl_landmarks_package.joblib"
hmm_metadata_path = MODEL_DIR / "hmm_metadata.json"

hmm_package = {
    "class_models": class_models,
    "scaler": scaler,
    "pca": pca,
    "label_encoder": label_encoder,
    "n_hmm_states": N_HMM_STATES,
    "n_pca_components": N_PCA_COMPONENTS,
}
joblib.dump(hmm_package, hmm_package_path)

metadata = {
    "model": "GaussianHMM one model per class",
    "dataset": DATASET_NAME,
    "selected_classes": list(label_encoder.classes_),
    "feature_type": "MediaPipe Holistic landmark sequences with StandardScaler and PCA",
    "num_sequences": int(len(sequence_labels)),
    "n_hmm_states": int(N_HMM_STATES),
    "n_pca_components": int(N_PCA_COMPONENTS),
    "accuracy": float(accuracy),
    "weighted_precision": float(precision),
    "weighted_recall": float(recall),
    "weighted_f1": float(f1),
}
hmm_metadata_path.write_text(json.dumps(metadata, indent=2), encoding="utf-8")

print("Saved HMM package:", hmm_package_path)
print("Saved metadata:", hmm_metadata_path)
```
