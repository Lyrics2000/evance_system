# Random Forest Baseline for WLASL Sign Recognition

This notebook builds a complete **Random Forest baseline** for sign recognition using WLASL loaded through FiftyOne.

It is meant for the comparison paper/thesis baseline, not as the final KSL healthcare model.

## What it does

1. Loads WLASL from FiftyOne/Hugging Face.
2. Selects a small manageable group of sign classes.
3. Extracts pose, hand, and face landmarks from videos using MediaPipe.
4. Converts every video into a fixed-length feature vector.
5. Trains a Random Forest classifier.
6. Evaluates the model using accuracy, precision, recall, F1-score, classification report, and confusion matrix.
7. Saves the model, label encoder, and metadata.

## Why Random Forest?

Random Forest is a traditional machine learning baseline. It is simple, explainable, and useful for comparison. Its weakness is that it does not naturally understand motion over time, so we summarize each video using statistics.
